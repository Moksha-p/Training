```python
# Create a class Library with attributes like name, address, and a
# list of books. Implement methods to add and remove books.
# ○ Example Test Case:
# ■ Input: name = "City Library", address = "123 Main St",
# books = []
# ■ Method Calls: add_book("Book1"), add_book("Book2"),
# remove_book("Book1")
# ■ Output: name: "City Library", address: "123 Main St",
# books: ["Book2"]

class Lirary:
    def __init__(self,name,address):
        self.name = name
        self.address = address
        self.books = []

    def add_book(self,title):
        self.books.append(title)
        print(f'{title} is added to library')
    
    def remove_book(self,title):
        for i in self.books:
            if i == title:
                self.books.remove(title)
            print(f'{title} is removed')
        else:
            print(f'No books found with name : {title}')

    def __str__(self):
        return f'name : {self.name}, address : {self.address}, books : {self.books}'

l = Lirary("City Library","123 Main st")
l.add_book("Book1")
l.add_book("Book2")
l.remove_book("Book1")
print(l)
```

```python
# 3. Create a class Book with attributes title, author, and price.
# Implement a method discount to apply a discount to the book's
# price.
# ○ Example Test Case:
# ■ Input: title = "Python 101", author = "John Doe",
# price = 29.99
# ■ Method Call: discount(0.1)
# ■ Output: title: "Python 101", author: "John Doe",
# price: 26.99

class Book:
    def __init__(self,title,author,price):
        self.title = title
        self.author = author
        self.price = price

    def discount(self,rate):
        self.price -= self.price * rate
        print(f'After Discount Value is : {self.price:.2f}')

    def detail(self):
        print(f'title : {self.title}, author : {self.author}, price : {self.price:.2f}')

b = Book("Python 101","John Doe",29.99)
b.detail()
b.discount(0.1)
b.detail()
```

```python
# Static Methods
# 7. Create a class MathUtils with a static method is_prime to check
# if a number is prime.
# ○ Example Test Case:
# ■ Input: number = 17
# ■ Method Call: is_prime(17)
# ■ Output: True

class MathUtils:
    
    @staticmethod
    def is_prime(n):
        if n <= 1:
            return False
        for i in range(2,n):
            if n % i == 0:
                return False
            
        return True
    
print(MathUtils.is_prime(17))
```

```python
# Inheritance
# 9. Create a base class Employee with attributes name and salary.
# Create a subclass Developer that adds an attribute
# programming_language.
# ○ Example Test Case:
# ■ Input: name = "Alice", salary = 70000,
# programming_language = "Python"
# ■ Output: name: "Alice", salary: 70000,
# programming_language: "Python"

class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary


class Developer(Employee):
    def __init__(self,name,salary,prog_language):
        super().__init__(name,salary)
        self.prog_language = prog_language

    

dev = Developer("Alice", 70000, "Python")
print(f'Name : {dev.name}, Salary : {dev.salary}, Programming_Language : {dev.prog_language}')
```

```python
# Create a base class Appliance with a method turn_on. Create a
# subclass WashingMachine that overrides the turn_on method.
# ○ Example Test Case:
# ■ Input: Appliance.turn_on(),
# WashingMachine(model="LG").turn_on()
# ■ Output: "Appliance is turned on", "Washing Machine LG
# is turned on"


class Appliance:

    def turn_on(self):
        print('Appliance is turned on')

class WashingMachine(Appliance):

    def __init__(self,model):
        self.model = model

    def turn_on(self):
        print(f"Washing Machine {self.model} is turned on")

appliance = Appliance()
machine = WashingMachine(model = "LG")
appliance.turn_on()
machine.turn_on()
```

```python
# Create a class PrivateData with private attributes username
# and password. Implement methods to get and set these attributes.
# ○ Example Test Case:
# ■ Input: username = "user1", password = "pass123"
# ■ Method Calls: get_username(),
# set_password("newpass123"), get_password()
# ■ Output: username: "user1", password: "newpass123"

class PrivateData:

    def __init__(self,username,password):
        self.__username = username
        self.__password = password

    def get_pass(self):
        return self.__password
    
    def get_username(self):
        return self.__username
    
    def set_pass(self, new_pass):
        self.__password = new_pass
        print(f'Password has been reset : {new_pass[:1]}{'*' * (len(new_pass) - 2)}{new_pass[-1]}')


p = PrivateData("user1","pass123")
print(p.get_username())
print(p.get_pass())
p.set_pass("pass1234")
```

```python
# 12. Create a class Account with private attributes account_number
# and balance. Implement methods to deposit and withdraw money.
# ○ Example Test Case:
# ■ Input: account_number = "987654321", balance = 5000
# ■ Method Calls: deposit(1500), withdraw(2000),
# get_balance()
# ■ Output: account_number: "987654321", balance: 4500

class Account:
    def __init__(self, account_number, balance):
        self.__account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds")

    def get_balance(self):
        return self.__balance

    def dis(self):
        return f'account_number: "{self.__account_number}", balance: {self.get_balance()}'

account = Account(account_number="987654321", balance=5000)
account.deposit(1500)
account.withdraw(2000)
print(account.dis())
```

```python
# Polymorphism
# 13. Create a function operate_vehicle that takes a vehicle object
# and calls its move method. Create classes Car and Bike that
# implement move differently.
# ○ Example Test Case:
# ■ Input: Car(model="Toyota").move(),
# Bike(model="Yamaha").move()
# ■ Output: "Toyota is driving", "Yamaha is riding"

class Vehicle:
    def __init__(self,model):
        self.model = model

    def move(self):
        pass


class Car(Vehicle):
    def move(self):
        print(f'{self.model} car is riding..!!')

class Bike(Vehicle):
    def move(self):
        print(f'{self.model} bike is riding..!!')

def operate_vehicle(obj):
    obj.move()

car = Car("BMW")
bike = Car("Yamha")

operate_vehicle(car)
operate_vehicle(bike)
```
