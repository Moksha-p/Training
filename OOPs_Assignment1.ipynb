```python
# 3 Implement a program that simulates a basic bank account using a BankAccount
# class.
# Example Test Case:
# ○ Input: account_number = "12345678", initial_balance = 1000
# ○ Output: account_number: "12345678", balance: 1000

class BankAccount:
    def __init__(self,acc_num,initial_balance):
        self.acc_num = acc_num
        self.initial_balance = initial_balance

    def __str__(self):
        return f'account_number: "{self.acc_num}", balance: {self.initial_balance}'

    def deposit(self,ammount):
        if ammount>0:
            self.initial_balance += ammount
            print(f"Amount: {ammount} \nNew balance: {self.initial_balance}")
        
    def withdraw(self,amount):
        if 0<amount<=self.initial_balance:
            self.initial_balance -= amount
            print(f"Withdraw amount: {amount} \nNew balance: {self.initial_balance}")

account = BankAccount("12345",15000)
print(account)
account.deposit(2000)
account.withdraw(2000)

```

```python
# 4 Write a Python program that uses a Rectangle class to calculate the area and
# perimeter of a rectangle.
# Example Test Case:
# ○ Input: length = 5, width = 3
# ○ Output: area: 15, perimeter: 16

class Rectangle:
    def __init__(self,l,b):
        self.l = l
        self.b = b

    def area(self):
        a = self.l * self.b
        return a
    
    def peri(self):
        perimeter = 2 * (self.l + self.b)
        return perimeter
    
r = Rectangle(5,3)
print(r.area())
print(r.peri())
```

```python
# 8 Implement a program that uses a Circle class to calculate the area and
# circumference of multiple circles.
# Example Test Case:
# ○ Input: radius = 4
# ○ Output: area: 50.27, circumference: 25.13
import math
class Circle:
    def __init__(self,r):
        self.r = r
    
    def area(self):
        a = math.pi * self.r * self.r
        return a

    def circum(self):
        c = 2 * math.pi * self.r
        return c
    
circle = Circle(4)
print(circle.area())
print(circle.circum())
```

```python
# 1 Create a base class Shape with methods to calculate area and perimeter, and
# derive classes Circle and Square.
# Example Test Case:
# ○ Input: Circle(radius=4), Square(side=5)
# ○ Output: Circle area: 50.27, Circle perimeter: 25.13, Square
# area: 25, Square perimeter: 20
import math
class Shape:
    def areas(self):
        raise NotImplementedError("Child class should implement this method")
    def perimeter(self):
        raise NotImplementedError("Child class should implement this method")
    
class Circle(Shape):
    def __init__(self,r):
        self.r = r
    def areas(self):
        return (math.pi*self.r)
    def peri(self):
        return math.pi*(2*self.r)
    
class Square(Shape):
    def __init__(self,side):
        self.side = side
    def areas(self):
        return (self.side*self.side)
    def peri(self):
        return 4*self.side
    
c = Circle(4)
s = Square(5)

print(f"Perimeter of circle: {c.peri():.2f} Area: {c.areas():.2f}")
print(f"Perimeter of square: {s.peri()} Area: {s.areas()}")
```

```python
# 2 Implement a class hierarchy to represent different types of employees (Manager,
# Engineer) with their attributes.
# Example Test Case:
# ○ Input: Manager(name="John Doe", department="Sales"),
# Engineer(name="Jane Smith", field="Software")
# ○ Output: Manager: John Doe, Department: Sales, Engineer: Jane
# Smith, Field: Software

class Employees:
    def __init__(self,name):
        self.name = name
    def __str__(self):
        return f'Employee: {self.name}'
    
class Manager(Employees):
    def __init__(self,name,department):
        super().__init__(name)
        self.department = department

    def __str__(self):
        return f"Name: {self.name} Department: {self.department}"
    
class Engineer(Employees):
    def __init__(self,name,field):
        super().__init__(name)
        self.field = field

    def __str__(self):
        return f"Name: {self.name} Department: {self.field}"
    
m = Manager("John Doe","Sales")
print(m)

e = Engineer("Jane Smith","Software")
print(e)
```

```python
# 3. Write a Python program that uses inheritance to represent a hierarchy of shapes
# (Triangle, Rectangle, etc.).
# # Example Test Case:
# Input: Triangle(base=5, height=3), Rectangle(length=4,
# width=2)
# ○ Output: Triangle area: 7.5, Rectangle area: 8
class Shapes:
    def area(self):
        raise NotImplementedError("Child class must have implement this method")
    
class Rectangle(Shapes):
    def __init__(self,l,w):
        self.l = l
        self.w = w
    def area(self):
        a = self.l *self.w
        return a
    
class Triangle(Shapes):
    def __init__(self,b,h):
        self.b = b
        self.h = h
    def area(self):
        a = 0.5 * self.b * self.h
        return a
    
r = Rectangle(l=4,w=2)
t = Triangle(b=5, h=3)

print(f"AREA:\nrectangle: {r.area()} \ntriangle: {t.area()}")
```

```python
# 4 Create a class hierarchy to represent different types of animals (Bird, Fish) with
# their own attributes and methods.
# Example Test Case:
# ○ Input: Bird(name="Parrot", can_fly=True),
# Fish(name="Goldfish", can_swim=True)
# ○ Output: Bird: Parrot, Can Fly: True, Fish: Goldfish, Can
# Swim: True
class Animal:
    def __init__(self,name):
        self.name = name
    def __str__(self):
        return f"Animal: {self.name}"
    
class Bird(Animal):
    def __init__(self,name,can_fly):
        super().__init__(name)
        self.can_fly = can_fly
    def __str__(self):
        return f"Bird(name={self.name},can fly={self.can_fly})"
    
class Fish(Animal):
    def __init__(self,name,can_swim):
        super().__init__(name)
        self.can_swim = can_swim
    def __str__(self):
        return f"Bird(name={self.name},can fly={self.can_swim})"
    
b = Bird(name="Parrot", can_fly=True)
f = Fish(name="Goldfish", can_swim=True)
print(b)
print(f)
```
