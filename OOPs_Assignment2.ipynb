### Class Definition and Object Instantiation

```python
# 1. Create a class Library with attributes like name, address, and a
# list of books. Implement methods to add and remove books.
# ○ Example Test Case:
# ■ Input: name = "City Library", address = "123 Main St",
# books = []
# ■ Method Calls: add_book("Book1"), add_book("Book2"),
# remove_book("Book1")
# ■ Output: name: "City Library", address: "123 Main St",
# books: ["Book2"]

class library:
    def __init__(self, name, address):
        self.name = name
        self.address = address
        self.bookList = []
    
    def add_books(self,book_name):
        self.bookList.append(book_name)
    
    def remove_book(self, book_name):
        self.bookList.remove(book_name)

    def display_books(self):
        return self.bookList
    
    def __str__(self):
        return f"Name: {self.name} Address: {self.address} Books: {self.bookList}"
        
library = library(name="City Library", address="123 Main St")
library.add_books("book1")    
library.add_books("book2")    
library.remove_book("book1")
print(library.display_books())
print(library)
```

### Instance Methods

```python
# 3. Create a class Book with attributes title, author, and price.
# Implement a method discount to apply a discount to the book's
# price.
# ○ Example Test Case:
# ■ Input: title = "Python 101", author = "John Doe",
# price = 29.99
# ■ Method Call: discount(0.1)
# ■ Output: title: "Python 101", author: "John Doe",
# price: 26.99

class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def discount(self,dis):
        self.price -= self.price * dis
        return f"Updated price after discount: {self.price}"
   
    def display_details(self):
        print(f'title: "{self.title}", author: "{self.author}", price: {self.price:.2f}')

book = Book(title="Python 101", author="John Doe", price=29.99)
book.display_details()
print(book.discount(0.1))  
book.display_details()
```

### Class Variables

```python
# 5. Create a class School with a class variable total_students and
# instance variables name and students. Implement a method to
# enroll students and update the total count.
# ○ Example Test Case:
# ■ Input: name = "Greenwood High", students = 300
# ■ Method Call: enroll_students(50)
# ■ Output: name: "Greenwood High", students: 350,
# total_students: 350

class School:
    total_students = 0

    def __init__(self,name,students):
        self.name = name
        self.students = students
        School.total_students += self.students

    def enroll_students(self,std):
        self.students += std
        School.total_students += std
    
    def display_school(self):
        return f"Name:{self.name} Students: {School.total_students}"
    
sch = School(name = "Greenwood High", students = 300)
print(sch.display_school())
sch.enroll_students(50)
print(sch.display_school())
```

### Static Methods

```python
# 7. Create a class MathUtils with a static method is_prime to check
# if a number is prime.
# ○ Example Test Case:
# ■ Input: number = 17
# ■ Method Call: is_prime(17)
# ■ Output: True

class MathUtils:
    @staticmethod
    def is_prime(num):
        for i in range(2,num):
            if num%i == 0:
                    
                return False
        return True
print(MathUtils.is_prime(16))
```

```python
# 9. Create a base class Employee with attributes name and salary.
# Create a subclass Developer that adds an attribute
# programming_language.
# ○ Example Test Case:
# ■ Input: name = "Alice", salary = 70000,
# programming_language = "Python"
# ■ Output: name: "Alice", salary: 70000,
# programming_language: "Python"

class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary


class Developer(Employee):
    def __init__(self,name,salary,programming_language):
        super().__init__(name,salary)
        self.programming_language = programming_language

    def display(self):
        print(f"Name: {self.name} Salary: {self.salary} Programming language: {self.programming_language}")

d = Developer(name = "Alice", salary = 70000, programming_language = "Python")
d.display()


    
```

### Encapsulation

```python
# 11. Create a class PrivateData with private attributes username
# and password. Implement methods to get and set these attributes.
# ○ Example Test Case:
# ■ Input: username = "user1", password = "pass123"
# ■ Method Calls: get_username(),
# set_password("newpass123"), get_password()
# ■ Output: username: "user1", password: "newpass123"

class PrivateData:
    def __init__(self,username,password):
        self.__username = username
        self.__password = password
    
    def get_username(self):
        return self.__username
    
    def get_password(self):
        return self.__password
    
    def set_username(self,newusername):
        self.__username = newusername

    def set_password(self,newpassword):
        self.__password = newpassword

pd = PrivateData(username = "user1", password = "pass123")
print(pd.get_username())
print(pd.get_password())
pd.set_username("user2")
pd.set_password("pass456")
print(pd.get_username())
print(pd.get_password())
```

```python
# 12. Create a class Account with private attributes account_number
# and balance. Implement methods to deposit and withdraw money.
# ○ Example Test Case:
# ■ Input: account_number = "987654321", balance = 5000
# ■ Method Calls: deposit(1500), withdraw(2000),
# get_balance()
# ■ Output: account_number: "987654321", balance: 4500

class Account:
    def __init__(self,account_number,balance):
        self.__account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds")

    def get_balance(self):
        return self.__balance

    def dis(self):
        return f'account_number: "{self.__account_number}", balance: {self.get_balance()}'

account = Account(account_number="987654321", balance=5000)
account.deposit(1500)
account.withdraw(2000)
print(account.dis())
```

```python
# 13. Create a function operate_vehicle that takes a vehicle object
# and calls its move method. Create classes Car and Bike that
# implement move differently.
# ○ Example Test Case:
# ■ Input: Car(model="Toyota").move(),
# Bike(model="Yamaha").move()
# ■ Output: "Toyota is driving", "Yamaha is riding"

class Vehicle:
    def __init__(self,model):
        self.model = model

    def move(self):
        pass

class Car(Vehicle):
    def move(self):
        print(f'{self.model} car is riding..!!')

class Bike(Vehicle):
    def move(self):
        print(f'{self.model} bike is riding..!!')

def operate_vehicle(obj):
    obj.move()

car = Car("Toyota")
bike = Car("Yamha")

operate_vehicle(bike)
```

### Abstraction

```python
# 15. Define an abstract class Transport with an abstract method
# travel. Create subclasses Bus and Train that implement the travel
# method.
# ○ Example Test Case:
# ■ Input: Bus(route="10A").travel(),
# Train(route="Express").travel()
# ■ Output: "Bus 10A is traveling", "Train Express is
# traveling"

from abc import ABC, abstractmethod

class Transport(ABC):
    def __init__(self, route):
        self.route = route

    @abstractmethod
    def travel(self):
        pass 

class Bus(Transport):
    def travel(self):
        print(f"Bus {self.route} is traveling")

class Train(Transport):
    def travel(self):
        print(f"Train {self.route} is traveling")

bus = Bus(route="10A")
bus.travel()  

train = Train(route="Express")
train.travel()  
```

```python
# 16. Define an abstract class Payment with an abstract method
# process. Create subclasses CreditCard and PayPal that implement
# the process method.
# ○ Example Test Case:
# ■ Input: CreditCard(number="1234").process(),
# PayPal(account="user@example.com").process()
# ■ Output: "Processing credit card payment for 1234",
# "Processing PayPal payment for user@example.com"

from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def process(self):
        pass  

class CreditCard(Payment):
    def __init__(self, number):
        self.number = number

    def process(self):
        print(f"Processing credit card payment for {self.number}")

class PayPal(Payment):
    def __init__(self, account):
        self.account = account

    def process(self):
        print(f"Processing PayPal payment for {self.account}")

credit_card_payment = CreditCard(number="1234")
credit_card_payment.process()  

paypal_payment = PayPal(account="user@example.com")
paypal_payment.process()  
```
