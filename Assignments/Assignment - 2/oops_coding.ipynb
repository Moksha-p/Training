```python
class Library:
    def __init__(self, name, address, books):
        self.name = name
        self.address = address
        self.books = books
    
    def add_book(self, title):
        self.books.append((title))
    
    def remove_book(self, title):
        if title in self.books:
            self.books.remove(title)
    
    def get_info(self):
        return f"Library Name: {self.name}, Address: {self.address}, Books: {self.books}"
    
if __name__ == '__main__':
    library = Library("City Library", "123 Main Street", [])
    library.add_book("Book1")
    library.add_book("Book2")
    library.remove_book('Book1')
    print(library.get_info())
```

```python
class House:
    def __init__(self, address, num_rooms, price):
        self.address = address
        self.num_rooms = num_rooms
        self.price = price
    
    def display_details(self):
        return f"Address: {self.address}, Number of Rooms: {self.num_rooms}, Price: {self.price}"

if __name__ == '__main__':
    house = House("4 Elm Street", 4, 350000)
    print(house.display_details())
```

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
    
    def discount(self, percent):
        discount_amount = self.price * (percent)
        self.price = self.price - discount_amount

    def display_info(self):
        return f"Title: {self.title}, Author: {self.author}, Price: {self.price}"

if __name__ == '__main__':
    book = Book("Python 101", "John Doe", 29.99)
    book.discount(0.1)
    print(book.display_info())
```

```python
class Restaurant():
    def __init__(self, name, cuisine_type, rating):
        self.name = name
        self.cuisine_type = cuisine_type
        self.rating = rating
    
    def update_rating(self, rating):
        self.rating = rating
    
    def display_info(self):
        return f"Name: {self.name}, Cuisine Type: {self.cuisine_type}, Rating: {self.rating}"

if __name__ == '__main__':
    restaurant = Restaurant("Sushi Place", "Japanese", 4.5)
    restaurant.update_rating(4.8)
    print(restaurant.display_info())
```

```python
class School:
    total_students = 0
    def __init__(self, name, students):
        self.name = name
        self.students = students
        School.total_students += students
    
    def enroll_students(self, students):
        self.students += students
        School.total_students += students
    
    def get_info(self):
        return f"School Name: {self.name}, Students: {self.students}, Total Students: {School.total_students}"
    
if __name__ == "__main__":
    school = School("Greenwood High School", 300)
    school.enroll_students(50)
    print(school.get_info())
```

```python
class MathUtils:
    def __init__(self):
        pass
    
    @staticmethod
    def is_prime(number):
        if number <= 1:
            return False
        for i in range(2, number):
            if number % i == 0:
                return False
        return True

if __name__ == "__main__":
    math_utils = MathUtils()
    print(math_utils.is_prime(17))
```

```python
class TemperatureConverter:
    def __init__(self):
        pass

    @staticmethod
    def celsius_to_fahrenheit(celsius):
        fahrenheit = (celsius * 9/5) + 32
        return fahrenheit

if __name__ == "__main__":
    converter = TemperatureConverter()
    print(converter.celsius_to_fahrenheit(25))
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

class Developer(Employee):
    def __init__(self, name, salary, programming_language):
        super().__init__(name, salary)
        self.programming_language = programming_language
    
    def get_info(self):
        return f"Name: {self.name}, Salary: {self.salary}, Programming Language: {self.programming_language}"
    
if __name__ == "__main__":
    developer = Developer("Alice", 70000, "Python")
    print(developer.get_info())
```

```python
class Account:
    def __init__(self, account_number, balance):
        self.__account_number = account_number
        self.__balance = balance
    
    def deposit(self, amount):
        self.__balance += amount
    
    def withdraw(self, amount):
        self.__balance -= amount
    
    def get_account_number(self):
        return self.__account_number
    
    def get_balance(self):
        return self.__balance

if __name__ == '__main__':
    account = Account('987654321', 5000)
    account.deposit(1500)
    #print(account.get_balance())
    account.withdraw(2000)

    print(f"account_number: {account.get_account_number()}, balance: {account.get_balance()}")
```

```python
class Car:
    def __init__(self, model):
        self.model = model
    
    def move(self):
        print(f"{self.model} is moving")

class Bike:
    def __init__(self, model):
        self.model = model
    
    def move(self):
        print(f"{self.model} is moving")

def operate_vehicle(vehicle):
    vehicle.move()

if __name__ == "__main__":
    car = Car("Toyota")
    bike = Bike("Yamaha")
    operate_vehicle(car)
    operate_vehicle(bike)
```

```python
def operate_device(device):
    device.operate()

class Laptop:
    def __init__(self, model):
        self.model = model
    
    def operate(self):
        print(f"{self.model} is operating")

class Smartphone:
    def __init__(self, model):
        self.model = model
    
    def operate(self):
        print(f"{self.model} is operating")

if __name__ == "__main__":
    laptop = Laptop("Dell")
    smartphone = Smartphone("iPhone")
    operate_device(laptop)
    operate_device(smartphone)
```

```python
from abc import ABC

class Transport(ABC):
    def travel(self):
        pass

class Bus(Transport):
    def __init__(self, route):
        self.route = route
    
    def travel(self):
        print(f"Bus {self.route} is traveling")

class Train(Transport):
    def __init__(self, route):
        self.route = route
    
    def travel(self):
        print(f"Train {self.route} is traveling")

if __name__ == "__main__":
    Bus('10A').travel()
    Train('Express').travel()
```

```python
class Payment(ABC):
    def process(self):
        pass

class CreditCard(Payment):
    def __init__(self, number):
        self.number = number
    
    def process(self):
        print(f"Processing credit card payment for {self.number}")

class PayPal(Payment):
    def __init__(self, account):
        self.account = account
    
    def process(self):
        print(f"Processing PayPal payment for {self.account}")

if __name__ == "__main__":
    CreditCard(1234).process()
    PayPal('user@example.com').process()
```
