```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b
    
    @staticmethod
    def subtract(a, b):
        return a - b
    
    @staticmethod
    def multiply(a, b):
        return a * b
    
    @staticmethod
    def divide(a, b):
        if b == 0:
            raise ZeroDivisionError("Cannot divide by zero")
        return a / b
    
if __name__ == "__main__":
    a = 10; b = 5
    cal = Calculator()
    print(f"Addition: {cal.add(a, b)}")
    print(f"Subtraction: {cal.subtract(a, b)}")
    print(f"Multiplication: {cal.multiply(a, b)}")
    print(f"Division: {cal.divide(a, b)}" )
```

```python
class StringUtils:
    @staticmethod
    def reverse_string(s):
        return s[::-1]
    
    @staticmethod
    def is_palindrome(s):
        return s == s[::-1]
    
if __name__ == '__main__':
    str_util = StringUtils()
    string = 'racecar'
    print(f"Reversed string: {str_util.reverse_string(string)}")
    print(f"Is Palindrome? {str_util.is_palindrome(string)}")
```

```python
class DateUtils:
    @staticmethod
    def is_leap_year(year):
        return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)
    @staticmethod
    def days_in_month(year, month):
        days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        if month == 2 and DateUtils.is_leap_year(year):
            days[month - 1] = 29

        return days[month - 1]
    
if __name__ == '__main__': 
    date_util = DateUtils()
    year = 2020; month = 2
    print(f"Is {year} a leap year? {date_util.is_leap_year(year)}")
    print(f"Number of days in {year}-{month:02d}: {date_util.days_in_month(year, month)}")
```
