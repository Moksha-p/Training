```python
# Q1] Write a Python program to find all the prime numbers between two given numbers.
# ■ Example Test Case:
# ■ Input: start = 10, end = 30
# ■ Expected Output: [11, 13, 17, 19, 23, 29]

start = 10
end = 30
primes = []

for i in range(start, end + 1):
    if i > 1:  
        flag = True
        for j in range(2, i):
            if i % j == 0:
                flag = False
                break
        if flag:
            primes.append(i)

print(primes)
```

```python
# Write a Python program to find the first n numbers that are both prime.
# ■ Example Test Case:
# ■ Input: n = 5
# ■ Expected Output: [2, 3, 5, 7, 11]

def find_first_n_primes(n):
    primes = []
    num = 2  
    while len(primes) < n:
        is_prime = True
        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
        num += 1  

    return primes

n = 5
print(find_first_n_primes(n))
```

```python
# Write a Python function to find the longest word in a given sentence.
# ■ Example Test Case:
# ■ Input: "The quick brown fox jumps over the lazy dog"
# ■ Expected Output: "jumps"
string = "The quick brown fox jumps over the lazy dog"
def longest(string):
    l1 = sorted(string.split(" "),key = len)
    return l1[-1]
longest(string)
```

```python
# Write a Python function to check if two given strings are anagrams of each other.
# ■ Example Test Case:
# ■ Input: str1 = "listen", str2 = "silent"
# ■ Expected Output: True
dictt1 = {}
dictt2 = {}
def anagrams(str1,str2):
    for i in str1:
        if i not in dictt1:
            dictt1[i] = 1
        else:
            dictt1[i] += 1
    for i in str2:
        if i not in dictt2:
            dictt2[i] = 1
        else:
            dictt2[i] += 1
    
    if dictt1 == dictt2:
        print("anagrams")
    else:
        print("not")

str1 = "listen"
str2 = "silent"
anagrams(str1,str2)
```

```python
# Create a program that evaluates a list of Boolean expressions and returns the count of
# True values.
# ■ Example Test Case:
# ■ Input: [True, False, True, True, False]
# ■ Expected Output: 3

l1 = [True, False, True, True, False]
def countTrues(l1):
    count = 0
    for i in l1:
        if i:
            count += 1
    return count
countTrues(l1)
```

```python
# Write a Python function to evaluate a complex Boolean expression represented as a string
# (e.g., "True and False or True").
# ■ Example Test Case:
# ■ Input: "True and False or True"
# ■ Expected Output: True
strr = "True and False or True"
def complexBoolean(strr):
    try:
        # Evaluate the boolean expression using eval
        result = eval(strr)
        return result
    except Exception as e:
        print("Error evaluating expression:", e)
        return None
complexBoolean(strr)
```

```python
# Create a function to merge two sorted tuples into one sorted tuple.
# ■ Example Test Case:
# ■ Input: tuple1 = (1, 3, 5), tuple2 = (2, 4, 6)
# ■ Expected Output: (1, 2, 3, 4, 5, 6)
def merge(t1,t2):
    lst = []
    for i in (t1,t2):
        for j in i:
            lst.append(j)
            sort = sorted(lst)
    return sort

t1 = (1, 3, 5)
t2 = (2, 4, 6)
merge(t1,t2)
```

```python
# Write a Python program to find all unique combinations of elements in a tuple that sum
# up to a given number.
# ■ Example Test Case:
# ■ Input: tuple = (2, 3, 5, 7), target = 10
# ■ Expected Output: [(3, 7), (2, 3, 5)]
ip = (2, 3, 5, 7)
target = 10
result = []

def backtrack(start,subset):
    if sum(subset) == target:
        result.append((subset[:]))

    for i in range(start,len(ip)):
        subset.append(ip[i])
        backtrack(i+1, subset)
        subset.pop()
backtrack(0,[])
print(result)

```

```python
# Write a Python program to merge multiple dictionaries and sum the values of common
# keys.
# ■ Example Test Case:
# ■ Input: dict1 = {'a': 1, 'b': 2}, dict2 = {'b': 3, 'c': 4}
# ■ Expected Output: {'a': 1, 'b': 5, 'c': 4}

dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}

for i in dict2:
    if i not in dict1:
        dict1[i] = dict2[i]
    else:
        dict1[i] += dict2[i]

print(dict1)
```

```python
# Write a Python function to find all subsets of a given set.
# ■ Example Test Case:
# ■ Input: set = {1, 2, 3}
# ■ Expected Output: [{}, {1}, {2}, {3}, {1, 2}, {1, 3}, {2,
# 3}, {1, 2, 3}]
def find_subsets(input_set):
    input_list = list(input_set)  
    result = []
    for i in range(2 ** len(input_list)):
        subset = [input_list[j] for j in range(len(input_list)) if (i & (1 << j))]
        result.append(subset)
    
    return result

input_set = {1, 2, 3}
subsets = find_subsets(input_set)
print(subsets)
```

```python
# Write a Python program to solve a quadratic equation (ax^2 + bx + c = 0) and classify the
# roots as real, complex, or equal.
# ■ Example Test Case:
# ■ Input: a = 1, b = -3, c = 2
# ■ Expected Output: Real roots: 2.0, 1.0
import math

def solve_quadratic(a, b, c):
    discriminant = b**2 - 4*a*c
    
    if discriminant > 0:
        root1 = (-b + math.sqrt(discriminant)) / (2 * a)
        root2 = (-b - math.sqrt(discriminant)) / (2 * a)
        return f"Real roots: {root1}, {root2}"
    elif discriminant == 0:
        root = -b / (2 * a)
        return f"Equal roots: {root}"
    else:
        real_part = -b / (2 * a)
        imag_part = math.sqrt(-discriminant) / (2 * a)
        return f"Complex roots: {real_part} + {imag_part}i, {real_part} - {imag_part}i"
solve_quadratic(1,-3,2)
```

```python
# ○ Write a Python function to generate the Fibonacci series up to a given number of terms.
# ■ Example Test Case:
# ■ Input: terms = 5
# ■ Expected Output: [0, 1, 1, 2, 3]

n = 5

def fibbo(n):
    if n == 1:
        print(0)
    elif n == 2:
        print(0,1)
    else:
        a = 0
        b = 1
        print(a,end = ' ')
        for i in range(n-1):
            print(b,end=' ')
            a,b = b, a+b
fibbo(n)
```

```python
# Write a Python program to find the longest increasing subsequence in a given list of
# numbers.
# ■ Example Test Case:
# ■ Input: [10, 22, 9, 33, 21, 50, 41, 60, 80]
# ■ Expected Output: [10, 22, 33, 50, 60, 80]

ip = [10, 22, 9, 33, 21, 50, 41, 60, 80]
op = [ip[0]]
for i in ip:
    if i > op[-1]:
        op.append(i)

print(op)
```

```python
# Write a Python program to find the greatest common divisor (GCD) of two numbers
# using the Euclidean algorithm.
# ■ Example Test Case:
# ■ Input: num1 = 48, num2 = 18
# ■ Expected Output: 6


a,b = 48,18

while b != 0:
    a,b = b, a % b

print(a)
```

```python
# /Write a Python program to implement a basic version of the Collatz conjecture.
# ■ Example Test Case:
# ■ Input: number = 13
# ■ Expected Output: [13, 40, 20, 10, 5, 16, 8, 4, 2, 1]
number = 13
def collatz_conjecture(number):
    sequence = [number]  
    while number != 1:
        if number % 2 == 0:
            number //= 2  
        else:
            number = 3 * number + 1  
        sequence.append(number) 
    return sequence
collatz_conjecture(number)
```
